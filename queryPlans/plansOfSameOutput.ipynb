### 2. Postgres 18

Comparisons of the query plans in pg18 of 10 runs

```python
import re
import difflib
from pathlib import Path
from collections import defaultdict

root = Path("/home/kristina/Pojects/Analysis-Scripts/plans_runs/20261012_172245")
runs = [f"run{i}" for i in range(1, 11)]
db, bench = "pg18", "tpch"

# lines that only carry runtime/measurement info, not plan shape
IGNORE = re.compile(
    r"^\s*(QUERY PLAN|-{5,}|Buffers:|Planning|Execution Time|JIT:|Functions:|Options:|Timing:"
    r"|Worker \d+:|Workers Launched|Sort Method|Buckets:|Rows Removed|Index Searches"
    r"|Heap Fetches|Peak Memory|Memory Usage|\(\d+ rows?\))"
)
# "(cost=... rows=... width=...)", "(actual rows=... loops=...)", "(never executed)"
ANNOT = re.compile(r"\s*\((?:cost|actual|never executed)[^)]*\)")

def normalize(lines, keep_costs=False):
    out = []
    for line in lines:
        if IGNORE.match(line):
            continue
        if not keep_costs:
            line = ANNOT.sub("", line)
        if line.strip():
            out.append(line.rstrip())
    return out

def load(run, query):
    return (root / run / db / bench / f"{query}.txt").read_text().splitlines()

# every query found in run1
queries = sorted(p.stem for p in (root / runs[0] / db / bench).glob("*.txt"))

for q in queries:
    groups = defaultdict(list)          # normalized plan -> runs that produced it
    for r in runs:
        try:
            plan = tuple(normalize(load(r, q)))
        except FileNotFoundError:
            print(f"{q}: missing in {r}")
            continue
        groups[plan].append(r)

    if len(groups) == 1:
        print(f"[same] {q}")
    else:
        print(f"[DIFF] {q}: {len(groups)} distinct plans")
        for plan, rs in groups.items():
            print("   ", ", ".join(rs))

        # show what differs between the first two variants
        (p1, r1), (p2, r2) = list(groups.items())[:2]
        for l in difflib.unified_diff(p1, p2, fromfile=r1[0], tofile=r2[0], lineterm=""):
            print("     ", l)
```

#### 3.b In.sql pg16 

In the test run, we found that one of the queries ran abnormally long. In.sql runs much longer than ay other queries, specifically in pg 16. This test runs quickly in version 18.

Here, we investigate the query plans as to why this is the case

```python
root16 = Path("/home/kristina/Pojects/Analysis-Scripts/plans_runs/20261012_210411") #only has 4 runs 
root18 = Path ("/home/kristina/Pojects/Analysis-Scripts/plans_runs/20261012_172245")# has 10 runs

exec = re.compile(r"Execution Time:\s*([\d.]+)\s*ms")

def exec_time(path):
    m = exec.search(path.read_text())
    return float(m.group(1)) if m else None

def run_dirs(root):
    return sorted((p for p in root.glob("run*") if p.is_dir()),
        key=lambda p: int(p.name[3:]))

def exec_times(root, db, query, bench="tpch"):
    times = {}
    for rd in run_dirs(root):
        f = rd / db / bench / f"{query}.txt"
        if f.exists():
            times[rd.name] = exec_time(f)
        else:
            print(f"{query}: missing in {rd.name}")
    return times
def show_exec_times(root, db, query):
    t = exec_times(root, db, query)
    print(f"{db} / {query}")
    for run, ms in t.items():
        print(f" {run:7} {ms} ms")
    vals = [v for v in t.values() if v is not None]
    if vals:
        print(f" mean {sum(vals)/len(vals):.3f} ms over {len(vals)} runs")

show_exec_times(root16, "pg16", "In")
show_exec_times(root18, "pg18", "In")
```

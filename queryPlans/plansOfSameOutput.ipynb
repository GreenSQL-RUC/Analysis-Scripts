```python
import re
import difflib
from pathlib import Path
from collections import defaultdict

root = Path("/home/kristina/Pojects/Analysis-Scripts/plans_runs/20261012_172245")
runs = [f"run{i}" for i in range(1, 11)]
db, bench = "pg18", "tpch"

# lines that only carry runtime/measurement info, not plan shape
IGNORE = re.compile(
    r"^\s*(QUERY PLAN|-{5,}|Buffers:|Planning|Execution Time|JIT:|Functions:|Options:|Timing:"
    r"|Worker \d+:|Workers Launched|Sort Method|Buckets:|Rows Removed|Index Searches"
    r"|Heap Fetches|Peak Memory|Memory Usage|\(\d+ rows?\))"
)
# "(cost=... rows=... width=...)", "(actual rows=... loops=...)", "(never executed)"
ANNOT = re.compile(r"\s*\((?:cost|actual|never executed)[^)]*\)")

def normalize(lines, keep_costs=False):
    out = []
    for line in lines:
        if IGNORE.match(line):
            continue
        if not keep_costs:
            line = ANNOT.sub("", line)
        if line.strip():
            out.append(line.rstrip())
    return out

def load(run, query):
    return (root / run / db / bench / f"{query}.txt").read_text().splitlines()

# every query found in run1
queries = sorted(p.stem for p in (root / runs[0] / db / bench).glob("*.txt"))

for q in queries:
    groups = defaultdict(list)          # normalized plan -> runs that produced it
    for r in runs:
        try:
            plan = tuple(normalize(load(r, q)))
        except FileNotFoundError:
            print(f"{q}: missing in {r}")
            continue
        groups[plan].append(r)

    if len(groups) == 1:
        print(f"[same] {q}")
    else:
        print(f"[DIFF] {q}: {len(groups)} distinct plans")
        for plan, rs in groups.items():
            print("   ", ", ".join(rs))

        # show what differs between the first two variants
        (p1, r1), (p2, r2) = list(groups.items())[:2]
        for l in difflib.unified_diff(p1, p2, fromfile=r1[0], tofile=r2[0], lineterm=""):
            print("     ", l)
```
